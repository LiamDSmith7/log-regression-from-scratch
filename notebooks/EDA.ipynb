```python
import seaborn as sns
import matplotlib.pyplot as plt

font_colour = "#1F2933"
point_colour = "#A44A3F"
distribution_colour = "#A44A3F"
background_colour = "#F7F4EE"

sns.set_theme(
    style="white",
    rc={
        "figure.facecolor": background_colour,
        "axes.facecolor": background_colour,
        "axes.edgecolor": font_colour,
        "axes.labelcolor": font_colour,
        "xtick.color": font_colour,
        "ytick.color": font_colour,
        "text.color": font_colour,
    }
)

pair_plot = sns.pairplot(
    df,
    plot_kws={
        "color": point_colour,
        "alpha": 0.75,
        "s": 35
    },
    diag_kws={
        "color": distribution_colour,
        "fill": True,
        "alpha": 0.8
    }
)

pair_plot.fig.patch.set_facecolor(background_colour)

pair_plot.savefig(
    "pairplot.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=background_colour
)

plt.show()
```

```python
import sys
from pathlib import Path

# Go up one directory from the notebook location
parent_dir = Path.cwd().parent

# Add that parent directory to Python's import search path
sys.path.append(str(parent_dir))

# Import model.py from the src folder
from src.model import LogisticRegressionScratch
from src.preprocessing import split_features_and_target, encode_labels, train_test_split, standardise_train_test, load_csv_with_headers
```

```python
headers, data = load_csv_with_headers('../data/Raisin_Dataset.csv')
```

```python
X, Y = split_features_and_target(data)
print(X)
print(Y)
```

```python
Y_encoded, Y_unique = encode_labels(Y)
```

```python
print(Y_encoded)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y_encoded, test_size=0.2)
```

```python
X_train_scaled, X_test_scaled, mean, std = standardise_train_test(X_train, X_test)
```

```python
print(y_train)
```

```python
import matplotlib.pyplot as plt

# Train model
model = LogisticRegressionScratch(
    learning_rate=0.01,
    n_iters=1000
)

model.fit(X_train_scaled, y_train)

# Presentation colour scheme
background_colour = "#F7F4EE"
text_colour = "#1F2933"
accent_colour = "#A44A3F"

# Plot loss over iterations
fig, ax = plt.subplots(figsize=(8, 5))

fig.patch.set_facecolor(background_colour)
ax.set_facecolor(background_colour)

ax.plot(
    range(len(model.losses)),
    model.losses,
    color=accent_colour,
    linewidth=2
)

ax.set_xlabel("Iteration", color=text_colour)
ax.set_ylabel("Binary Cross-Entropy Loss", color=text_colour)
ax.set_title("Objective Function During Training", color=text_colour)

ax.tick_params(colors=text_colour)

for spine in ax.spines.values():
    spine.set_color(text_colour)

ax.grid(True, alpha=0.2)


plt.savefig(
    "objective_function_training.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(model.errors, color="#A44A3F")
plt.xlabel("Iteration")
plt.ylabel("Mean Absolute Error")
plt.title("Prediction Error During Training")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(model.bias_history, color="#A44A3F")
plt.xlabel("Iteration")
plt.ylabel("Bias")
plt.title("Bias During Training")
plt.show()
```

```python
import numpy as np 

weights_history = np.array(model.weights_history)

plt.figure(figsize=(8, 5))

for i in range(weights_history.shape[1]):
    plt.plot(weights_history[:, i], label=f"Weight {i}")

plt.xlabel("Iteration")
plt.ylabel("Weight value")
plt.title("Weights During Training")
plt.legend()
plt.show()
```
